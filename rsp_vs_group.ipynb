# rsp_vs_group

このNotebookは `group_agents/rsp_group_<n>.py` を読み込み、堅牢化した対戦エンジン `rsp_group_engine.py` で団体戦を実行します。

- 提出ファイルは対戦前にAST検査されます。
- 各エージェントは別プロセスで実行されます。
- 1ラウンドでは両者の手を先に取得し、その後にだけ相手の手を通知します。
- 対戦ごとにエージェントを新規生成するため、前の相手との状態は持ち越されません。


```python
from pathlib import Path

import pandas as pd

from rsp_group_engine import discover_group_files, run_group_tournament_from_dir, print_tournament_summary

base_dir = Path("group_agents")
num_match = 10000
team_size = 3
timeout_seconds = 1.0
```

```python
groups = discover_group_files(base_dir)
pd.DataFrame([{"group": g.group, "file": str(g.path)} for g in groups])
```

```python
%%time
result = run_group_tournament_from_dir(
    base_dir,
    num_match=num_match,
    team_size=team_size,
    timeout_seconds=timeout_seconds,
)
print_tournament_summary(result)
```

```python
group_scores = result["group_scores"]
group_scores
```

```python
scores = result["scores"]
scores
```

```python
detail_rows = []
for detail in result["details"]:
    for side in ("agent1", "agent2"):
        row = {
            "match": f"{detail['group1']} vs {detail['group2']}",
            "agent": detail[side],
            "agent_index": detail["agent_index"],
        }
        row.update(detail[f"{side}_result"])
        detail_rows.append(row)

details = pd.DataFrame(detail_rows)
details
```
